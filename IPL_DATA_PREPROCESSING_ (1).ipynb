# **Most winning teams**
---



```python
wins = matches['winner'].value_counts().head(10)
```

```python
print(wins)
```

```python
import matplotlib.pyplot as plt
```

```python
wins.plot(kind='bar',figsize=(10,5),title='MOST WINNING TEAMS IN IPL')
plt.xlabel('winner')
plt.ylabel('no of wins')
plt.show()
```

# ***Top run scorers ***

```python
highest_runs = deliveries[['batter','batsman_runs']].groupby('batter').sum().sort_values('batsman_runs',ascending=False ).head(10)
```

```python
print(highest_runs)
```

```python
highest_runs.plot(kind='bar',figsize=(10,5),title='HIGHEST IPL SCORE BATSMAN')
plt.xlabel=('batsman_runs')
plt.ylabel=('batter')
plt.show()
```

# **Top wicket takers **

```python
most_wickets = deliveries[['bowler','is_wicket']].groupby('bowler').sum().sort_values('is_wicket',ascending=False).head(10)
```

```python
print(most_wickets)
```

```python
most_wickets.plot(kind='bar',figsize=(10,5),title='MOST WICKETS TAKEN')
plt.xlabel=('is_wicket')
plt.ylabel=('bowler')
plt.show()
```

# **Toss vs match winner **

```python
match_winner = (matches['toss_winner'] == matches['winner']).value_counts()
```

```python
print(match_winner)
```

```python
match_winner.plot(kind='pie',figsize=(10,5),title='IF A TEAM WON THE TOSS DID IT WON THE MATCH')
plt.show()
```
